## CO3: Data Cleaning and Preparation

Data cleaning is an essential step in exploratory data analysis because missing values, duplicate records, inconsistent column names, and invalid entries can affect the reliability of results. In this section, the news article dataset is checked for missing values and duplicates, and numerical and categorical columns are examined to determine whether preprocessing is required.

```python
import pandas as pd
import numpy as np

clean_df = df.copy()

print("Missing values in each column:")
display(clean_df.isnull().sum()[clean_df.isnull().sum() > 0])

print("Total missing values:", clean_df.isnull().sum().sum())
print("Duplicate rows:", clean_df.duplicated().sum())
print("Duplicate column names:", clean_df.columns.duplicated().sum())

clean_df["shares"] = clean_df["shares"].fillna(clean_df["shares"].median())

print("Missing shares after filling:", clean_df["shares"].isnull().sum())
```

```python
clean_df = df.copy()

sample_df = clean_df[["shares", "n_tokens_title", "n_tokens_content"]].head(10).copy()

sample_df = sample_df.rename(
    columns={
        "shares": "article_shares",
        "n_tokens_title": "headline_words",
        "n_tokens_content": "article_words"
    }
)

print("Renamed columns:")
display(sample_df)

clean_df["popularity_category"] = pd.cut(
    clean_df["shares"],
    bins=[-1, 1000, 5000, 10000, np.inf],
    labels=["Low", "Moderate", "High", "Very High"]
)

print("Popularity categories:")
display(clean_df["popularity_category"].value_counts().sort_index())
```

```python
shares = df["shares"]

Q1 = shares.quantile(0.25)
Q3 = shares.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(shares < lower_bound) | (shares > upper_bound)]
filtered_df = df[(shares >= lower_bound) & (shares <= upper_bound)]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of outliers:", len(outliers))
print("Rows before filtering:", len(df))
print("Rows after filtering:", len(filtered_df))

display(outliers[["shares"]].head(10))
```

```python
sample_df = df[["shares", "n_tokens_title", "n_tokens_content"]].head(20).copy()

sample_df["article_id"] = range(1, len(sample_df) + 1)
sample_df["popularity_group"] = np.where(
    sample_df["shares"] >= 2000, "Popular", "Less Popular"
)

multi_df = sample_df.set_index(["popularity_group", "article_id"])

print("Hierarchical index:")
display(multi_df)

print("Sorted hierarchical index:")
display(multi_df.sort_index(level="popularity_group"))

print("Summary by popularity group:")
display(
    multi_df.groupby(level="popularity_group")["shares"].agg(
        ["count", "mean", "median"]
    )
)
```

## CO4: Data Aggregation and Group Operations

Data aggregation helps summarize large datasets by grouping records according to common characteristics. In this section, news articles are grouped by category and publication day to compare their popularity. Multiple aggregation functions are applied to calculate the number of articles, mean shares, median shares, and maximum shares for each group.

```python

# Identify news category columns
category_cols = [
    "data_channel_is_lifestyle",
    "data_channel_is_entertainment",
    "data_channel_is_bus",
    "data_channel_is_socmed",
    "data_channel_is_tech",
    "data_channel_is_world",
    "data_channel_is_",
]

# Use category labels if available in the dataset
category_map = {
    "data_channel_is_lifestyle": "Lifestyle",
    "data_channel_is_entertainment": "Entertainment",
    "data_channel_is_bus": "Business",
    "data_channel_is_socmed": "Social Media",
    "data_channel_is_tech": "Technology",
    "data_channel_is_world": "World",
    "data_channel_is_": "Other",
}

available_categories = [
    col for col in category_cols if col in df.columns
]

# Create a category label for each article
if available_categories:
    df["category"] = "Other / Unclassified"
    for col in available_categories:
        df.loc[df[col] == 1, "category"] = category_map[col]

    category_summary = df.groupby("category")["shares"].agg(
        article_count="count",
        mean_shares="mean",
        median_shares="median",
        maximum_shares="max"
    ).round(2)

    print("Popularity by news category:")
    display(category_summary.sort_values("median_shares", ascending=False))
else:
    print("Category indicator columns were not found. Check df.columns.")

# Weekday popularity, using the dataset's weekday indicator columns
weekday_cols = [
    "weekday_is_monday",
    "weekday_is_tuesday",
    "weekday_is_wednesday",
    "weekday_is_thursday",
    "weekday_is_friday",
    "weekday_is_saturday",
    "weekday_is_sunday"
]

weekday_names = [
    "Monday", "Tuesday", "Wednesday", "Thursday",
    "Friday", "Saturday", "Sunday"
]

available_weekdays = [col for col in weekday_cols if col in df.columns]

if len(available_weekdays) == 7:
    weekday_summary = pd.DataFrame({
        "day": weekday_names,
        "article_count": [int(df[col].sum()) for col in weekday_cols],
        "median_shares": [
            df.loc[df[col] == 1, "shares"].median()
            for col in weekday_cols
        ],
        "mean_shares": [
            df.loc[df[col] == 1, "shares"].mean()
            for col in weekday_cols
        ]
    }).round(2)

    print("\nPopularity by publication day:")
    display(weekday_summary)
else:
    print("Some weekday indicator columns were not found.")
```

```python
weekday_cols = [
    "weekday_is_monday",
    "weekday_is_tuesday",
    "weekday_is_wednesday",
    "weekday_is_thursday",
    "weekday_is_friday",
    "weekday_is_saturday",
    "weekday_is_sunday"
]

weekday_names = [
    "Monday", "Tuesday", "Wednesday", "Thursday",
    "Friday", "Saturday", "Sunday"
]

weekday_frames = []

for col, day in zip(weekday_cols, weekday_names):
    day_data = df.loc[df[col] == 1, ["shares"]].copy()
    day_data["publication_day"] = day
    weekday_frames.append(day_data)

weekday_data = pd.concat(weekday_frames, ignore_index=True)

weekday_summary = weekday_data.groupby("publication_day").agg(
    article_count=("shares", "count"),
    mean_shares=("shares", "mean"),
    median_shares=("shares", "median")
).reindex(weekday_names).round(2)

display(weekday_summary)
```

## CO5: Plotting and Visualization

Data visualization helps identify patterns, distributions, and relationships that may not be apparent from numerical summaries alone. Matplotlib and Seaborn are used to create histograms, bar plots, and scatter plots to investigate news article popularity. Appropriate titles, axis labels, legends, and figure layouts are used to improve clarity, and the resulting visualizations are saved for further analysis and reporting.

```python

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set_theme(style="whitegrid")

# Create a 2 × 2 visualization layout
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Histogram: article popularity on a log scale
sns.histplot(
    x=np.log1p(df["shares"].dropna()),
    bins=40,
    ax=axes[0, 0]
)
axes[0, 0].set_title("Distribution of Article Popularity")
axes[0, 0].set_xlabel("log(1 + shares)")
axes[0, 0].set_ylabel("Number of Articles")

# 2. Bar plot: median shares by category, if available
if "category" in df.columns:
    category_medians = (
        df.groupby("category")["shares"]
        .median()
        .sort_values(ascending=False)
    )
    sns.barplot(
        x=category_medians.index,
        y=category_medians.values,
        ax=axes[0, 1]
    )
    axes[0, 1].set_title("Median Shares by Category")
    axes[0, 1].set_xlabel("Category")
    axes[0, 1].set_ylabel("Median Shares")
    axes[0, 1].tick_params(axis="x", rotation=35)
else:
    axes[0, 1].text(
        0.5, 0.5, "Category summary not available",
        ha="center", va="center"
    )
    axes[0, 1].set_axis_off()

# 3. Scatter plot: headline length versus popularity
sns.scatterplot(
    data=df,
    x="n_tokens_title",
    y="shares",
    alpha=0.25,
    s=15,
    ax=axes[1, 0]
)
axes[1, 0].set_title("Headline Length vs Article Popularity")
axes[1, 0].set_xlabel("Headline Length (words)")
axes[1, 0].set_ylabel("Shares")

# 4. Histogram: article content length
sns.histplot(
    data=df,
    x="n_tokens_content",
    bins=40,
    ax=axes[1, 1]
)
axes[1, 1].set_title("Distribution of Article Length")
axes[1, 1].set_xlabel("Article Length (words)")
axes[1, 1].set_ylabel("Number of Articles")

plt.tight_layout()
plt.savefig("news_popularity_visualizations.png",
            dpi=150, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select relevant numerical features from the news dataset
features = [
    "shares",
    "n_tokens_title",
    "n_tokens_content",
    "num_hrefs",
    "num_imgs",
    "num_videos",
    "global_sentiment_polarity",
    "title_sentiment_polarity"
]

# Keep only columns available in the dataset
features = [col for col in features if col in df.columns]

# Calculate Pearson correlation
correlation_matrix = df[features].corr(method="pearson")

# Plot the correlation heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Pearson Correlation Between News Article Features")
plt.tight_layout()
plt.show()
```

## Final Findings and Conclusion

This project presents an exploratory data analysis of the Online News Popularity dataset to investigate the factors associated with the popularity of online news articles. The dataset contains 39,644 articles and 61 original features, including article content characteristics, publication information, sentiment measures, and the number of shares.

NumPy was used for numerical computations, array manipulation, indexing, slicing, and universal functions. Pandas was used for data inspection, filtering, sorting, data cleaning, missing-value handling, transformation, merging, reshaping, and descriptive statistics. Grouping and aggregation techniques helped compare article popularity across news categories and publication days. Matplotlib and Seaborn were used to create histograms, bar plots, line plots, scatter plots, density plots, box plots, FacetGrid visualizations, and correlation heatmaps.

The analysis revealed that article popularity is highly right-skewed. The median number of shares was 1,400, while the mean was approximately 3,395, indicating that a relatively small number of highly shared articles increased the average. Among the publication days, Saturday had the highest median number of shares at 2,000, followed by Sunday at 1,900. The median for weekdays was approximately 1,400, compared with 1,900 for weekends.

Differences were also observed across news categories. Other or unclassified articles had a mean of approximately 5,945 shares, while Social Media articles had a median of 2,100 shares. These results demonstrate that average and median values can provide different perspectives on article popularity.

The correlation analysis showed that the relationships between individual article features and popularity were generally weak. The number of hyperlinks had a Spearman correlation of approximately 0.090 with shares, followed by the number of images at approximately 0.083 and positive sentiment at approximately 0.080. These associations do not establish causation and should not be interpreted as evidence that any single feature directly increases popularity.

Overall, this project demonstrates how exploratory data analysis, statistical summaries, data wrangling, and visualization can be combined to examine a real-world dataset. The findings provide useful insights into online news-sharing patterns while highlighting the importance of data quality, appropriate statistical measures, and careful interpretation of results.
